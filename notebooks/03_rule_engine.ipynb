# 03 — Explainable rule engine

This notebook evaluates the configured R001–R012 rules and compares broad Version 1 thresholds with the more contextual Version 2 thresholds.

```python
import sqlite3
import pandas as pd
from pipeline.rules.engine import evaluate_rules

conn = sqlite3.connect('../data/generated/fraud_command_center.db')
transactions = pd.read_sql('SELECT * FROM transactions', conn)
customers = pd.read_sql('SELECT * FROM customers', conn)
features = pd.read_sql('SELECT * FROM transaction_features', conn)
v1 = evaluate_rules(transactions, features, customers, version='V1')
v2 = evaluate_rules(transactions, features, customers, version='V2')
v2.results.head()
```

```python
print(v1.results.groupby('rule_id').size().sort_values(ascending=False))
print(v2.results.groupby('rule_id').size().sort_values(ascending=False))
print(v2.triggered_matrix['rule_score'].describe())
```
