# 04 — Model evaluation

Logistic Regression is the interpretable baseline. Isolation Forest is an additional anomaly signal. The split is chronological, not random.

```python
import sqlite3
import pandas as pd
from pipeline.ml.evaluate import evaluate_models

conn = sqlite3.connect('../data/generated/fraud_command_center.db')
transactions = pd.read_sql('SELECT * FROM transactions', conn)
features = pd.read_sql('SELECT * FROM transaction_features', conn)
rule_score = pd.read_sql('SELECT transaction_id, SUM(rule_score) AS rule_score FROM rule_results GROUP BY transaction_id', conn)
signals, report, metric_rows = evaluate_models(transactions, features, rule_score)
report['metrics']
```
