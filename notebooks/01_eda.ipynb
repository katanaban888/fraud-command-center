# 01 — Exploratory data analysis

This notebook inspects the reproducible synthetic dataset, distributions, missing values and class imbalance. It uses the same pipeline seed as the application.

```python
from pathlib import Path
import pandas as pd
import sqlite3

DB = Path('../data/generated/fraud_command_center.db')
conn = sqlite3.connect(DB)
transactions = pd.read_sql('SELECT * FROM transactions', conn)
customers = pd.read_sql('SELECT * FROM customers', conn)
transactions.head()
```

## Volume, distributions and data quality

```python
transactions.info()
transactions[['amount','balance_before','balance_after']].describe()
transactions.isna().mean().sort_values(ascending=False).head(10)
```

```python
fraud_rate = transactions['is_fraud'].mean()
print(f'Fraud rate: {fraud_rate:.2%}')
print(transactions['fraud_scenario'].value_counts())
print(transactions.groupby('channel')['is_fraud'].mean().sort_values(ascending=False))
```

## Conclusions

The dataset is intentionally imbalanced. All fraud-rate and volume conclusions should be recomputed after the generator seed or configuration changes; no result is copied into the dashboard manually.
